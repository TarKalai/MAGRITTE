# Synthetic Signature Fraud — BPI Challenge 2012

Generates `data/processed/bpic_2012_signature.parquet`: BPIC 2012 with a **structural behavioural-signature**
fraud injected into fraud trace. This notebook mirrors `scripts/regenerate_signature_dataset.py`, the
canonical generator of the published parquet.

**Why not random Nolle anomalies?** We first tried Nolle/april control-flow anomalies (skip/insert/reorder).
They proved **unlearnable by a supervised classifier**: each fraud trace gets a *different, random* mutation, so
there is no recurring pattern to learn (a positive control — injecting an unmistakable sentinel — hit ROC-AUC 1.0,
confirming the pipeline is correct and the issue was the signal). Nolle anomalies target *unsupervised* detectors;
supervised classifiers need a **consistent** signature.

**The structural signature.** The fraud is a **rare consecutive transition** — a back-loop `a → b → a` between two
**common** activities — injected into the region **before the surrogate transaction** (`O_ACCEPTED+COMPLETE`), the
region the model observes. Insertion positions are random, so a few fraud traces receive no motif and others several.
    The injection probability **escalates**
toward the transaction, so detectability rises as more of the prefix is observed.

**Difficulty knob — class overlap.** A deterministic signature would be detected at ~1.0. To keep the task
non-trivial we also inject the same motif into a fraction (`CONTAM_FRAC`) of *legit* traces (kept labelled normal):
the pattern is *indicative but not exclusive*, capping peak detectability. Proxy for 'account takeover follows a
recognizable script that legit users occasionally resemble'.

## 1. Setup — lightweight `april` import

We only need `EventLog`/`Event`; pre-seed bare packages and stub `april.fs` to skip april's TensorFlow-heavy
`__init__` and the `arrow` dependency. This notebook lives in `notebooks/`, so the repo root is the parent of the
working directory and `april` is vendored under `lib/april`.

```python
import sys, types
from pathlib import Path
from collections import Counter
import numpy as np
import pandas as pd

REPO = Path.cwd().parent            # notebooks/ -> repo root
APRIL = REPO / 'lib' / 'april'      # vendored april subset
for name, sub in [('april', ''), ('april.processmining', 'processmining'), ('april.generation', 'generation')]:
    m = types.ModuleType(name)
    m.__path__ = [str(APRIL / sub)] if sub else [str(APRIL)]
    sys.modules[name] = m
fs = types.ModuleType('april.fs')
fs.EVENTLOG_DIR = REPO / 'local' / '.out' / 'eventlogs'
fs.EVENTLOG_CACHE_DIR = REPO / 'local' / '.out' / '.cache' / 'eventlogs'
sys.modules['april.fs'] = fs
from april.processmining.log import EventLog
from april.processmining.event import Event
```

## 2. Configuration

`ESC_MAX` (escalation strength) and `CONTAM_FRAC` (class overlap) are the knobs that set the signature and its difficulty. The motif itself is **chosen from the data** in step 4 (the rarest transition between common activities).

```python
SEED = 42
FRAUD_RATE = 0.05                  # fraction of eligible (surrogate-bearing) traces turned into fraud
SURROGATE = 'O_ACCEPTED+COMPLETE'  # surrogate transaction; signature injected in the prefix before it
ESC_MAX = 0.15                     # per-position injection prob escalates 0 -> ESC_MAX toward the transaction
MIN_PREFIX = 4                     # need >= this many events before the surrogate
CONTAM_FRAC = 0.20                 # legit traces also showing the pattern (caps peak AUC)
XES = REPO / 'data' / 'raw' / 'BPI_Challenge_2012.xes'
OUT = REPO / 'data' / 'processed' / 'bpic_2012_signature.parquet'
EPOCH = pd.Timestamp('2011-10-01 00:00:00', tz='UTC')
```

## 3. Load the log

```python
log = EventLog.from_xes(str(XES))
print(f'{len(log.cases)} traces | {len(set(log.unique_activities))} activities')
```

## 4. Choose the structural motif

Pick the **rarest consecutive transition `a → b` between two common activities** (each appearing in >= 30% of
traces). The injected motif is the back-loop `[a, b, a]` (transitions `a→b` and `b→a`).

```python
ntr = len(log.cases)
presence = Counter(); bigram = Counter()
for case in log:
    names = [e.name for e in case.events]
    for a in set(names):
        presence[a] += 1
    for a, b in zip(names, names[1:]):
        bigram[(a, b)] += 1
common = [a for a, c in presence.items() if c / ntr >= 0.30 and a != SURROGATE]   # appear in >=30% of traces
# pick the ordered common pair with the FEWEST consecutive occurrences in the log (rarest transition)
cand = sorted(((bigram.get((a, b), 0), a, b) for a in common for b in common if a != b))
_, A, B = cand[0]
MOTIF = [A, B, A]   # a->b->a back-loop (transitions a->b and b->a)
print(f"common activities: {len(common)} | chosen rare transition: {A}  ->  {B}  "
      f"(consecutive in normal: {bigram.get((A, B), 0)} times)")
print(f"motif (structural loop): {MOTIF}")
```

## 5. Empirical inter-event delta pool

Deltas are measured from the original timestamps and later resampled so every trace (fraud or normal) gets realistic, **label-neutral** timing — the signal is purely the structural signature, never the timing.

```python
deltas_pool = []
for case in log:
    ts = pd.to_datetime([e.timestamp for e in case.events], utc=True, errors='coerce')
    d = ts.to_series().diff().dt.total_seconds().dropna().values
    deltas_pool.extend(d[d > 0].tolist())
deltas_pool = np.array(deltas_pool)
print(f'delta pool: {len(deltas_pool):,} positive inter-event gaps')
```

## 6. Inject the signature (escalating)

The motif is inserted with probability escalating from 0 toward the surrogate, so more of the observed prefix means more chances the abnormal transition has appeared (rising earliness curve).

```python
def inject(case, cut):
    """Inject the structural loop (consecutive a->b->a) with probability escalating toward the transaction."""
    prefix, suffix = list(case.events[:cut]), case.events[cut:]
    P = max(len(prefix) - 1, 1)
    new_prefix = []
    for i, ev in enumerate(prefix):
        new_prefix.append(ev)
        if np.random.uniform(0, 1) < ESC_MAX * (i / P):
            new_prefix.extend(Event(name=m, timestamp=None) for m in MOTIF)
    case.events = new_prefix + suffix
```

## 7. Run injection over all traces

Each eligible trace (surrogate present, prefix >= `MIN_PREFIX`) becomes fraud with prob `FRAUD_RATE`; otherwise it is a contamination candidate with prob `CONTAM_FRAC` (motif injected but kept labelled normal — a hard negative).

```python
np.random.seed(SEED)
case_label, cut_after = [], []
for case in log:
    names = [e.name for e in case.events]
    cut = names.index(SURROGATE) if SURROGATE in names else None
    if cut is not None and cut >= MIN_PREFIX and np.random.uniform(0, 1) <= FRAUD_RATE:
        inject(case, cut); case_label.append('fraud')
    elif cut is not None and cut >= MIN_PREFIX and np.random.uniform(0, 1) <= CONTAM_FRAC:
        inject(case, cut); case_label.append('normal')
    else:
        case_label.append('normal')
    nm = [e.name for e in case.events]
    cut_after.append(nm.index(SURROGATE) if SURROGATE in nm else np.nan)
```

## 8. Re-derive timestamps (label-neutral) and export

Timestamps are rebuilt for every trace by resampling deltas from the empirical pool — identical treatment of fraud/normal, so timing carries no label information.

```python
rows = []
for case, lab in zip(log, case_label):
    n = len(case.events)
    if n == 0:
        continue
    d = np.random.choice(deltas_pool, size=max(n - 1, 0))
    ts = EPOCH + pd.to_timedelta(np.concatenate([[0.0], np.cumsum(d)]), unit='s')
    for e, t in zip(case.events, ts):
        rows.append((str(case.id), e.name, t, lab, 'signature' if lab == 'fraud' else 'normal'))

df = pd.DataFrame(rows, columns=['case_id', 'activity', 'timestamp', 'label', 'anomaly_type'])
OUT.parent.mkdir(parents=True, exist_ok=True)
df.to_parquet(OUT, index=False, engine='pyarrow', compression='snappy')

per = df.groupby('case_id').agg(label=('label', 'first'))
nf = (per.label == 'fraud').sum()
ln = df.groupby('case_id').size()
fr_idx = set(per[per.label == 'fraud'].index)
fr_cut = [c for sid, c in zip(per.index, cut_after) if sid in fr_idx and not np.isnan(c)]
print(f'traces={len(per)} | fraud={nf} ({nf/len(per)*100:.2f}%) | events={len(df)} -> {OUT.name}')
print('fraud transaction position (post-injection) pct:',
      {p: int(np.percentile(fr_cut, p)) for p in (25, 50, 75)})
print('eligible (>=L):', {L: int((ln >= L).sum()) for L in (30, 40, 50)})
```
